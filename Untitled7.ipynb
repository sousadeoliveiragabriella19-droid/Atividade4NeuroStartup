```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt
import keras
from keras import layers
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay
import ipywidgets as widgets
from ipywidgets import interact

SEED = 42
keras.utils.set_random_seed(SEED)

CLASSES = ["Camiseta/top", "Calça", "Pulôver", "Vestido", "Casaco",
           "Sandália", "Camisa", "Tênis", "Bolsa", "Bota"]

(x_a, y_a), (x_b, y_b) = keras.datasets.fashion_mnist.load_data()
X = np.concatenate([x_a, x_b])          # 70.000 fotos 28x28, tons de cinza (0 a 255)
y = np.concatenate([y_a, y_b])          # categoria de 0 a 9
df = pd.DataFrame({"categoria": [CLASSES[i] for i in y]})
print("Fotos:", X.shape, "| Respostas:", y.shape)

contagem = df["categoria"].value_counts().reindex(CLASSES)
ax = contagem.plot(kind="bar", figsize=(9, 4), color="steelblue")
ax.set_title("Quantidade de fotos por categoria"); ax.set_ylabel("Nº de fotos")
plt.xticks(rotation=45, ha="right"); plt.tight_layout(); plt.show()

fig, eixos = plt.subplots(2, 5, figsize=(11, 5))
for c, ax in enumerate(eixos.flat):
    ax.imshow(X[y == c].mean(axis=0), cmap="gray_r")
    ax.set_title(CLASSES[c]); ax.axis("off")
fig.suptitle("A 'foto média' de cada categoria")
plt.tight_layout(); plt.show()

X = X.astype("float32") / 255.0   # dica do professor: pixels de 0–255 para 0–1 (substitui o StandardScaler)

X_resto, X_teste, y_resto, y_teste = train_test_split(
    X, y, test_size=0.2, random_state=SEED, stratify=y)              # 🔒 cofre
X_treino, X_valid, y_treino, y_valid = train_test_split(
    X_resto, y_resto, test_size=0.25, random_state=SEED, stratify=y_resto)

print(f"Treino: {len(X_treino)} | Validação: {len(X_valid)} | Teste (cofre): {len(X_teste)}")

# versões "achatadas" (784 colunas) para os modelos do scikit-learn
X_treino_v = X_treino.reshape(len(X_treino), -1)
X_valid_v  = X_valid.reshape(len(X_valid), -1)

# Modelo simples (Aula 05): regressão logística nos 784 pixels (com max_iter aumentado para 1000)
baseline = LogisticRegression(max_iter=1000)
baseline.fit(X_treino_v, y_treino)
acc_base_valid = accuracy_score(y_valid, baseline.predict(X_valid_v))

# ⭐ Bônus: mais um modelo para comparar
arvore = DecisionTreeClassifier(max_depth=15, random_state=SEED).fit(X_treino_v, y_treino)
acc_arv_valid = accuracy_score(y_valid, arvore.predict(X_valid_v))

print(f"Regressão logística – validação: {acc_base_valid:.1%}")
print(f"Árvore de decisão   – validação: {acc_arv_valid:.1%}")
```
